```python
import pandas as pd
```

```python
prefix='https://github.com/DataTalksClub/nyc-tlc-data/releases/download/yellow/'
dtype = {
    "VendorID": "Int64",
    "passenger_count": "Int64",
    "trip_distance": "float64",
    "RatecodeID": "Int64",
    "store_and_fwd_flag": "string",
    "PULocationID": "Int64",
    "DOLocationID": "Int64",
    "payment_type": "Int64",
    "fare_amount": "float64",
    "extra": "float64",
    "mta_tax": "float64",
    "tip_amount": "float64",
    "tolls_amount": "float64",
    "improvement_surcharge": "float64",
    "total_amount": "float64",
    "congestion_surcharge": "float64"
}
parse_dates = [
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime"
]
```

```python
df=pd.read_csv(prefix+'/yellow_tripdata_2021-01.csv.gz',dtype=dtype,parse_dates=parse_dates)
```

```python
df.shape
```

```python
df.head(10)
```

```python
from sqlalchemy import create_engine
con=create_engine('postgresql://root:root@localhost:5432/ny_taxi')
```

```python
df.head(0).to_sql(name='yellow_taxi_data',con=con,if_exists='replace')
```

```python
pd.read_sql('select * from yellow_taxi_data where 1=0',con=con)
```
